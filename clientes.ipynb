```python
import dotenv
import pandas as pd
import duckdb
import requests
import json
from SQL.script_extracao import QUERY_CLIENTE


HEADERS = {
    "x-api-key": dotenv.get_key(".env", "API_KEY"),
    "content-type": "application/json"
}

URL = dotenv.get_key(".env", "URL_CLIENTES")
PAGE_SIZE = 100


def varrer_api(url, headers, count=PAGE_SIZE):
    resultados = []
    start = 0
    while True:
        params = {"start": start, "count": count}
        resp = requests.get(url, headers=headers, params=params)
        resp.raise_for_status()
        data = resp.json()
        items = data.get("items", [])
        if not items:
            break
        resultados.extend(items)
        start += count
    return resultados

dados = varrer_api(URL, HEADERS)
with open("CLIENTES.json", "w", encoding="utf-8") as f:
    json.dump({"items": dados}, f, ensure_ascii=False, indent=4)

#print(item)
result = duckdb.sql(QUERY_CLIENTE)
print(result)
```
